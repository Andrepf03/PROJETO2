**Célula 2 – criar o split e verificar:**

```python
X_train, y_train, X_val, y_val, X_test, y_test = get_splits()

for name, X, y in [("train", X_train, y_train), ("val", X_val, y_val), ("test", X_test, y_test)]:
    print(f"{name:5s} X: {X.shape} {X.dtype} | min={X.min():.1f} max={X.max():.1f} | y: {y.shape}")
```

**Célula 3 – contagem por classe em cada partição:**

```python
split_counts = pd.DataFrame({
    "class": CLASS_NAMES,
    "train": np.bincount(y_train, minlength=10),
    "val": np.bincount(y_val, minlength=10),
    "test": np.bincount(y_test, minlength=10),
})
print(split_counts.to_string(index=False))
split_counts.to_csv("results/tables/02_split_counts.csv", index=False)
```

**Célula 4 – teste de reprodutibilidade:**

```python
X_train2, y_train2, X_val2, y_val2, _, _ = get_splits()
print("Treino igual?    ", np.array_equal(X_train, X_train2) and np.array_equal(y_train, y_train2))
print("Validação igual? ", np.array_equal(X_val, X_val2) and np.array_equal(y_val, y_val2))
del X_train2, y_train2, X_val2, y_val2  
```
